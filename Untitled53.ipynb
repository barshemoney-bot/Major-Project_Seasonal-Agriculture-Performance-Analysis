# **Phase 1: Data Cleaning & Preparation**

# **Handle** **missing** **values** (**e**.**g**., **Rainfall_mm**, **Soil_Moisture_pct**, **and** **Yield_Tonnes_Ha** **have** **slight** **missing** **data** **in** **the** 4,**000**-**row** **dataset**).

# **Validate** **data** **types** **and** **standardize** **categorical** **variables** **like** **Season**, **Irrigation_Method**, **and** **Crop**.

```python
import pandas as pd

# Load dataset
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')

# 1. Missing Values
missing_info = df.isnull().sum()
missing_cols = missing_info[missing_info > 0]

# 2. Duplicates
duplicates = df.duplicated().sum()

# 3. Categorical Check
seasons = df['Season'].unique()
irrigation = df['Irrigation_Method'].unique()

print("Missing Values:")
print(missing_cols)
print(f"\nTotal Duplicates: {duplicates}")
print(f"\nUnique Seasons: {seasons}")
print(f"Unique Irrigation Methods: {irrigation}")

# Perform basic cleaning for demonstration and summary
# Fill missing Rainfall, Soil Moisture, and Yield with median
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

print(f"\nMissing values after cleaning: {df.isnull().sum().sum()}")
```

The data cleaning and preparation phase is complete. Here is the breakdown of the initial data state and the steps taken to prepare it for analysis:

Missing Value Identification: Out of 4,000 records, there were 48 missing values in Rainfall_mm, 40 in Soil_Moisture_pct, and 32 in Yield_Tonnes_Ha.

Imputation Strategy: Since these are continuous numerical variables that can be affected by outliers (like extreme weather events), all missing values were imputed using the median of their respective columns. The dataset now has zero missing values.

Duplicate Check: The dataset was scanned for duplicate rows, and 0 duplicates were found. Every record represents a unique entry.

Categorical Validation: The core categorical columns were verified for consistency. No structural errors or misspellings were found.

Seasons: Limited cleanly to three standard categories: 'Kharif', 'Rabi', and 'Zaid'.

Irrigation Methods: Cleanly categorized into 'Drip', 'Flood', 'Rainfed', and 'Sprinkler'.

# **Phase 2: Exploratory Data Analysis (EDA)Map out distributions of environmental conditions (temperature, rainfall, soil moisture) across different seasons.  Analyze the variation of resources used (water, fertilizers, pesticides) grouped by season and region.**

# **Phase 4: Synthesis & RecommendationsIdentify significant seasonal patterns or unusual anomalies.  Develop evidence-based recommendations for seasonal agricultural planning.**

```python
import pandas as pd

# Load and prepare data
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

# Synthesis Table 1: Overall Seasonal Efficiency
seasonal_efficiency = df.groupby('Season').agg(
    Avg_Yield_Tonnes_Ha=('Yield_Tonnes_Ha', 'mean'),
    Avg_Water_Used_m3=('Water_Used_m3', 'mean'),
    Avg_Profit_INR=('Profit_INR', 'mean')
).round(2)

# Synthesis Table 2: High vs Low Profit Crops in Zaid (The toughest season)
zaid_crops = df[df['Season'] == 'Zaid'].groupby('Crop').agg(
    Avg_Profit=('Profit_INR', 'mean'),
    Avg_Cost=('Total_Cost_INR', 'mean')
).sort_values(by='Avg_Profit', ascending=False).round(2)

print("--- Seasonal Efficiency Summary ---")
print(seasonal_efficiency)
print("\n--- Zaid Season Crop Performance ---")
print(zaid_crops)
```

The final phase of the project synthesizes the observed seasonal differences and provides evidence-based recommendations for agricultural planning.

Synthesis of Key Findings Seasonal Profitability Gap: Agricultural performance changes drastically across seasons, with Kharif generating the highest average profit (178,914.65 INR) and Zaid resulting in an average net loss (-24,804.82 INR).  Resource Inefficiency in Zaid: The Zaid season requires the highest water usage (6,419.89 m³) to combat high temperatures and low rainfall but yields the lowest production (4.64 Tonnes/Ha).  Crop-Specific Economic Viability: Profitability is heavily dependent on crop selection rather than general seasonal conditions. Cash crops like Sugarcane and Chilli maintain high profitability across all seasons, while staples like Rice, Maize, and Wheat operate at a consistent loss, peaking in the Zaid season.  

Data-Driven Recommendations for Agricultural PlanningRestrict Water-Intensive Staples in Zaid: Farmers should avoid planting Rice and Maize during the Zaid season, as these crops require massive water inputs (averaging over 6,400 m³) and result in severe financial losses (e.g., Rice averages a -227,239.34 INR loss in Zaid).Prioritize Cash Crops for Financial Stability: To offset the high baseline costs of farming (averaging over 500,000 INR per season), prioritize the cultivation of Sugarcane and Chilli, which remain heavily profitable regardless of the seasonal environmental variations.  Optimize Kharif for Maximum Yield: Since Kharif offers the most favorable natural environmental conditions (highest rainfall and soil moisture), resources should be heavily allocated to this season to maximize total annual yield and create a financial buffer for the more difficult Zaid season.  

```python
import pandas as pd

# 1. Load and prepare data
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')

# Handle missing values using medians
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

# 2. Synthesis Table 1: Overall Seasonal Efficiency
# Investigating how agricultural performance varies across seasons
seasonal_efficiency = df.groupby('Season').agg(
    Avg_Yield_Tonnes_Ha=('Yield_Tonnes_Ha', 'mean'),
    Avg_Water_Used_m3=('Water_Used_m3', 'mean'),
    Avg_Cost_INR=('Total_Cost_INR', 'mean'),
    Avg_Profit_INR=('Profit_INR', 'mean')
).round(2)

print("--- Seasonal Efficiency Summary ---")
print(seasonal_efficiency)

# 3. Synthesis Table 2: High vs Low Profit Crops in Zaid
# Identifying unusual or unexpected seasonal patterns (highest losses)
zaid_crops = df[df['Season'] == 'Zaid'].groupby('Crop').agg(
    Avg_Yield=('Yield_Tonnes_Ha', 'mean'),
    Avg_Cost=('Total_Cost_INR', 'mean'),
    Avg_Profit=('Profit_INR', 'mean')
).sort_values(by='Avg_Profit', ascending=False).round(2)

print("\n--- Zaid Season Crop Performance ---")
print(zaid_crops)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load and Prepare Data
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

# Set visualization style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(18, 5))

# Chart 1: Average Profit by Season
plt.subplot(1, 3, 1)
sns.barplot(data=df, x='Season', y='Profit_INR', errorbar=None, palette='viridis', hue='Season', legend=False)
plt.title('Average Profit by Season', fontsize=14, fontweight='bold')
plt.ylabel('Profit (INR)')
plt.xlabel('Season')

# Chart 2: Water Usage vs. Rainfall across Seasons
plt.subplot(1, 3, 2)
sns.scatterplot(data=df, x='Rainfall_mm', y='Water_Used_m3', hue='Season', palette='deep', alpha=0.6)
plt.title('Rainfall vs Applied Water Usage', fontsize=14, fontweight='bold')
plt.ylabel('Water Used (m³)')
plt.xlabel('Rainfall (mm)')

# Chart 3: Crop Profitability in the Zaid Season
plt.subplot(1, 3, 3)
zaid_data = df[df['Season'] == 'Zaid']
sns.barplot(data=zaid_data, x='Profit_INR', y='Crop', errorbar=None, palette='coolwarm_r', hue='Crop', legend=False)
plt.title('Crop Profitability (Zaid Season Only)', fontsize=14, fontweight='bold')
plt.ylabel('Crop Type')
plt.xlabel('Profit (INR)')

plt.tight_layout()
plt.show()
```

# **Phase 6 :Conclusion & Recommendations Summarize the evidence-based insights: Kharif is the most profitable season, while staple crops in the Zaid season lead to heavy financial losses.  Provide the final recommendation: Restrict water-intensive staples during Zaid and prioritize resilient cash crops like Sugarcane and Chilli to maintain year-round financial stability.**

```python
import pandas as pd

# Load data
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')

# Calculate supporting metrics
kharif_profit = df[df['Season'] == 'Kharif']['Profit_INR'].mean()
zaid_staples = df[(df['Season'] == 'Zaid') & (df['Crop'].isin(['Rice', 'Wheat', 'Maize']))]['Profit_INR'].mean()
cash_crops = df[df['Crop'].isin(['Sugarcane', 'Chilli'])]['Profit_INR'].mean()

# Define conclusion text
conclusion = f"""
CONCLUSION & RECOMMENDATIONS

Evidence-Based Insights:
1. Kharif is the most profitable season (Average Profit: {kharif_profit:,.2f} INR).
2. Cultivating staple crops in the Zaid season leads to heavy financial losses (Average Loss: {zaid_staples:,.2f} INR).

Final Recommendations:
1. Restrict water-intensive staples during the Zaid season to mitigate extreme resource costs.
2. Prioritize resilient cash crops like Sugarcane and Chilli (Average Year-Round Profit: {cash_crops:,.2f} INR) to maintain financial stability.
"""

# Print to console
print(conclusion)

# Save to a text file for final project submission
with open('Final_Recommendations.txt', 'w') as file:
    file.write(conclusion)
```
